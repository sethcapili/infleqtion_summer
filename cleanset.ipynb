## Figures

Three views of the 30x5 results, built from `Dataset` above.

1. **Mean score by strategy** - the headline number.
2. **Verdict mix** - *where* each strategy fails, not just how much.
3. **Circuit x strategy heatmap** - which circuits resist every strategy.

Colors follow the dataviz reference palette (light mode): categorical slots 1-6
for verdicts, single-hue blue ramp for the heatmap.

```python
# === Figures: prompting-strategy evaluation over 30 circuits ==================
# Palette = dataviz reference instance (light mode): categorical slots 1-6 for
# verdicts, documented single-hue blue ramp for the sequential heatmap.

import json, collections
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap

# --- palette -----------------------------------------------------------------
SURFACE, INK, INK2, MUTED, GRID, AXIS = (
    "#fcfcfb", "#0b0b0b", "#52514e", "#898781", "#e1e0d9", "#c3c2b7")
CAT = ["#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#e87ba4", "#008300",
       "#4a3aa7", "#e34948"]
BLUE_RAMP = ["#cde2fb", "#b7d3f6", "#9ec5f4", "#86b6ef", "#6da7ec", "#5598e7",
             "#3987e5", "#2a78d6", "#256abf", "#1c5cab", "#184f95", "#104281",
             "#0d366b"]
SEQ = LinearSegmentedColormap.from_list("blue_seq", BLUE_RAMP)

plt.rcParams.update({
    "figure.facecolor": SURFACE, "axes.facecolor": SURFACE,
    "savefig.facecolor": SURFACE,
    "font.family": "sans-serif",
    "font.sans-serif": ["Segoe UI", "DejaVu Sans", "sans-serif"],
    "text.color": INK, "axes.labelcolor": INK2,
    "xtick.color": MUTED, "ytick.color": MUTED,
    "axes.edgecolor": AXIS, "axes.linewidth": 1.0,
    "xtick.bottom": False, "ytick.left": False,
    "pdf.fonttype": 42, "ps.fonttype": 42,   # embed real text, not outlines
})


def save(fig, stem):
    """PDF is the paper deliverable; PNG is a preview."""
    fig.savefig(f"{stem}.pdf", bbox_inches="tight")
    fig.savefig(f"{stem}.png", dpi=200, bbox_inches="tight")

# --- data --------------------------------------------------------------------
try:
    Dataset  # already defined when run inside cleanset.ipynb
except NameError:
    _nb = json.load(open("cleanset.ipynb", encoding="utf-8"))
    _ns = {}
    exec("".join(_nb["cells"][2]["source"]), _ns)
    Dataset = _ns["Dataset"]

STRATS = ["1", "2", "3", "4", "5"]
SLABEL = {"1": "S1\nDirect", "2": "S2\nStricter", "3": "S3\nFew-shot",
          "4": "S4\nJSON", "5": "S5\nJSON+Repair"}

circuits = []
for r in Dataset:
    if r["circuit_type"] not in circuits:
        circuits.append(r["circuit_type"])

grid = {(r["circuit_type"], r["prompting_strategy"]): r for r in Dataset}
scores = np.array([[grid[(c, s)]["score"] for s in STRATS] for c in circuits],
                  dtype=float)

# `fully_correct` and `success` are the same outcome recorded under two labels.
VMAP = {"fully_correct": "success"}
def verdict(r):
    return VMAP.get(r["verdict"], r["verdict"])

VNICE = {"success": "Success", "execution_error": "Execution error",
         "invalid_syntax": "Invalid syntax", "wrong_gates": "Wrong gates",
         "wrong_statevector": "Wrong statevector",
         "wrong_qubit_count": "Wrong qubit count",
         "no_circuit_found": "No `circuit` variable"}


def _strip(ax, grid_axis="y"):
    """Recessive chrome: hairline grid on one axis, no box."""
    for side in ("top", "right", "left"):
        ax.spines[side].set_visible(False)
    ax.spines["bottom"].set_color(AXIS)
    ax.set_axisbelow(True)
    ax.grid(axis=grid_axis, color=GRID, linewidth=1.0)


# === Figure 1: mean score per strategy ========================================
def fig_means(stem="fig1_mean_score"):
    means = scores.mean(axis=0)
    fig, ax = plt.subplots(figsize=(7.5, 4.4))
    # Single series -> single hue; the x axis carries identity, so no legend.
    ax.bar(range(5), means, width=0.62, color=CAT[0], zorder=3)
    for i, m in enumerate(means):
        ax.text(i, m + 2.2, f"{m:.1f}", ha="center", va="bottom",
                fontsize=12, fontweight="bold", color=INK)
    ax.set_xticks(range(5))
    ax.set_xticklabels([SLABEL[s] for s in STRATS], fontsize=10, color=INK2)
    ax.set_ylim(0, 108)
    ax.set_yticks([0, 25, 50, 75, 100])
    ax.set_ylabel("Mean validator score (0–100)", fontsize=10)
    ax.text(0, 1.02, "Mean score across 30 circuits, by prompting strategy",
            transform=ax.transAxes, fontsize=10, color=MUTED, va="bottom")
    _strip(ax)
    fig.tight_layout()
    save(fig, stem)
    return fig


# === Figure 2: verdict composition per strategy ===============================
def fig_verdicts(stem="fig2_verdict_mix"):
    counts = {s: collections.Counter(verdict(r) for r in Dataset
                                     if r["prompting_strategy"] == s)
              for s in STRATS}
    totals = collections.Counter()
    for s in STRATS:
        totals.update(counts[s])
    # Success at the bottom of the stack, failure modes by overall frequency.
    order = ["success"] + sorted((k for k in totals if k != "success"),
                                 key=lambda k: -totals[k])

    fig, ax = plt.subplots(figsize=(8.6, 4.8))
    bottom = np.zeros(5)
    for i, v in enumerate(order):
        vals = np.array([counts[s].get(v, 0) for s in STRATS], dtype=float)
        # 2px surface gap between stacked segments.
        ax.bar(range(5), vals, bottom=bottom, width=0.62, color=CAT[i],
               label=VNICE.get(v, v), edgecolor=SURFACE, linewidth=2, zorder=3)
        for x, (val, b) in enumerate(zip(vals, bottom)):
            if val >= 3:  # selective direct labels, not one on every segment
                ax.text(x, b + val / 2, f"{int(val)}", ha="center", va="center",
                        fontsize=10, fontweight="bold", color="#ffffff",
                        zorder=4)
        bottom += vals

    ax.set_xticks(range(5))
    ax.set_xticklabels([SLABEL[s] for s in STRATS], fontsize=10, color=INK2)
    ax.set_ylim(0, 30)
    ax.set_yticks([0, 10, 20, 30])
    ax.set_ylabel("Circuits (of 30)", fontsize=10)
    ax.text(0, 1.02, "Where each strategy breaks down, by verdict",
            transform=ax.transAxes, fontsize=10, color=MUTED, va="bottom")
    _strip(ax)
    ax.legend(frameon=False, fontsize=9, loc="upper left",
              bbox_to_anchor=(1.01, 1.0), labelcolor=INK2)
    fig.tight_layout()
    save(fig, stem)
    return fig


# === Figure 3: circuit x strategy heatmap =====================================
def fig_heatmap(stem="fig3_heatmap"):
    # Hardest circuits first, so the rows that resist every strategy lead.
    idx = np.argsort(scores.mean(axis=1), kind="stable")
    m = scores[idx]
    labels = [circuits[i] for i in idx]

    fig, ax = plt.subplots(figsize=(7.2, 10.5))
    ax.imshow(m, cmap=SEQ, vmin=0, vmax=100, aspect="auto")

    for i in range(m.shape[0]):
        for j in range(m.shape[1]):
            v = m[i, j]
            ax.text(j, i, f"{int(v)}", ha="center", va="center", fontsize=9,
                    fontweight="bold",
                    color="#ffffff" if v >= 60 else INK)

    ax.set_xticks(range(5))
    ax.set_xticklabels([SLABEL[s].replace("\n", " ") for s in STRATS],
                       fontsize=9, color=INK2)
    ax.xaxis.set_ticks_position("top")
    ax.set_yticks(range(len(labels)))
    ax.set_yticklabels(labels, fontsize=9, color=INK2)
    # 2px surface gap between cells.
    ax.set_xticks(np.arange(-0.5, 5, 1), minor=True)
    ax.set_yticks(np.arange(-0.5, len(labels), 1), minor=True)
    ax.grid(which="minor", color=SURFACE, linewidth=2)
    ax.tick_params(which="minor", length=0)
    for side in ("top", "right", "left", "bottom"):
        ax.spines[side].set_visible(False)
    fig.tight_layout()
    save(fig, stem)
    return fig

# --- render ---
fig_means()
fig_verdicts()
fig_heatmap()
```

```python
"""Figure 4: the Strategy-5 pipeline, and a real repair-loop trace.

(a) schematic of prompt -> LLM -> extract -> validate -> {dataset | repair}
(b) the Bit-flip Error Correction run from walkthrough.ipynb, with the actual
    check values recomputed from repair_validator cell 21.
"""
import matplotlib
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch, Rectangle

SURFACE, INK, INK2, MUTED = "#fcfcfb", "#0b0b0b", "#52514e", "#898781"
GRID, AXIS = "#e1e0d9", "#c3c2b7"
BLUE, BLUE_SOFT = "#2a78d6", "#cde2fb"
GOOD, CRIT = "#0ca30c", "#d03b3b"

plt.rcParams.update({
    "figure.facecolor": SURFACE, "axes.facecolor": SURFACE,
    "savefig.facecolor": SURFACE,
    "font.family": "sans-serif",
    "font.sans-serif": ["Segoe UI", "DejaVu Sans", "sans-serif"],
    "text.color": INK,
    "pdf.fonttype": 42, "ps.fonttype": 42,   # embed real text, not outlines
})

CHECKS = ["syntax", "execution", "qubits", "gates",
          "measurement", "statevector", "simulation"]
A1 = dict(zip(CHECKS, [True, False, False, False, False, False, False]))
A2 = dict(zip(CHECKS, [True] * 7))


def box(ax, x, y, w, h, text, fc=SURFACE, ec=AXIS, fs=8.5, weight="normal",
        tc=INK, lw=1.1, radius=0.12):
    ax.add_patch(FancyBboxPatch(
        (x - w / 2, y - h / 2), w, h,
        boxstyle=f"round,pad=0,rounding_size={radius}",
        facecolor=fc, edgecolor=ec, linewidth=lw, zorder=2))
    ax.text(x, y, text, ha="center", va="center", fontsize=fs,
            color=tc, zorder=3, weight=weight, linespacing=1.45)


def arrow(ax, p0, p1, color=AXIS, lw=1.3, style="-|>", rad=0.0, ls="-"):
    ax.add_patch(FancyArrowPatch(
        p0, p1, arrowstyle=style, mutation_scale=11, linewidth=lw,
        color=color, zorder=1, linestyle=ls,
        connectionstyle=f"arc3,rad={rad}",
        shrinkA=1, shrinkB=1))


def panel_a(ax):
    ax.set_xlim(0, 10); ax.set_ylim(0, 10); ax.axis("off")
    cx = 6.0
    ys = [9.35, 7.75, 6.15, 4.30, 2.35, 0.62]

    box(ax, cx, ys[0], 6.2, 0.95,
        "Prompt\n" + r"JSON schema + user request", fs=8.5)
    box(ax, cx, ys[1], 6.2, 0.95, "LLM  ·  Qwen3-8B", fc=BLUE_SOFT, ec=BLUE,
        fs=9, weight="bold")
    box(ax, cx, ys[2], 6.2, 0.95,
        "Extract code\nJSON field  ·  fenced block", fs=8.5)
    box(ax, cx, ys[3], 6.2, 1.55,
        "Validator  —  7 checks\n"
        "syntax · execution · qubits · gates\nmeasurement · statevector · simulation",
        fs=8.2)
    # decision
    box(ax, cx, ys[4], 6.2, 0.95, "all 7 checks pass?", fs=8.5, ec=MUTED)
    box(ax, cx, ys[5], 6.2, 0.95,
        "Dataset entry\n" + r"score = 100 × k/7", fs=8.5, ec=GOOD, lw=1.4)

    for a, b in zip(ys[:-1], ys[1:]):
        arrow(ax, (cx, a - 0.48 if a != ys[3] else a - 0.78),
              (cx, b + (0.48 if b != ys[3] else 0.78)))

    ax.text(cx + 0.18, (ys[4] + ys[5]) / 2 + 0.02, "yes", fontsize=7.8,
            color=GOOD, ha="left", va="center", weight="bold")

    # repair path: decision -> back up to the LLM
    lx = 1.45
    arrow(ax, (cx - 3.1, ys[4]), (lx, ys[4]), style="-")
    arrow(ax, (lx, ys[4]), (lx, ys[1]), style="-", color=CRIT)
    arrow(ax, (lx, ys[1]), (cx - 3.1, ys[1]), color=CRIT)
    ax.text(cx - 3.25, ys[4] + 0.30, "no", fontsize=7.8, color=CRIT,
            ha="right", va="bottom", weight="bold")
    # single line -- two rotated lines collide at this width
    ax.text(lx - 0.30, (ys[1] + ys[4]) / 2,
            "repair prompt  ·  error categories",
            fontsize=7.8, color=CRIT, ha="center", va="center", rotation=90)

    ax.text(0.0, 9.95, "(a)  pipeline", fontsize=10, color=INK,
            ha="left", va="top", weight="bold")


def attempt_block(ax, y, label, checks, verdict, score, reason, code_note,
                  ok):
    ax.text(0.0, y + 0.92, label, fontsize=8.6, color=INK, weight="bold",
            ha="left", va="center")
    ax.text(9.9, y + 0.92, f"score {score}", fontsize=8.6,
            color=GOOD if ok else CRIT, weight="bold", ha="right", va="center")

    # check squares
    n = len(CHECKS)
    w, gap = 0.86, 0.55
    total = n * w + (n - 1) * gap
    x0 = (10 - total) / 2
    for i, k in enumerate(CHECKS):
        x = x0 + i * (w + gap)
        passed = checks[k]
        ax.add_patch(Rectangle((x, y), w, 0.5,
                               facecolor=GOOD if passed else SURFACE,
                               edgecolor=GOOD if passed else CRIT,
                               linewidth=1.2, zorder=2))
        # Segoe UI has no check/cross glyph; DejaVu does.
        ax.text(x + w / 2, y + 0.25, "✓" if passed else "✕",
                ha="center", va="center", fontsize=8, family="DejaVu Sans",
                color="#ffffff" if passed else CRIT, zorder=3, weight="bold")
        ax.text(x + w / 2, y - 0.20, k, ha="center", va="top", fontsize=6.3,
                color=MUTED, rotation=0)

    ax.text(0.0, y - 0.74, code_note, fontsize=7.6, color=INK2,
            ha="left", va="center", family="monospace")
    ax.text(9.9, y - 0.74, f"{verdict}" + (f"  ·  {reason}" if reason else ""),
            fontsize=7.6, color=GOOD if ok else CRIT, ha="right", va="center")


def panel_b(ax):
    ax.set_xlim(0, 10); ax.set_ylim(0, 10); ax.axis("off")
    ax.text(0.0, 9.95, "(b)  repair loop on Bit-flip Error Correction",
            fontsize=10, color=INK, ha="left", va="top", weight="bold")
    # Bit-flip and Superdense Coding are the only two circuits where S5 reaches
    # 100 and no other strategy does -- verified against the merged Dataset.
    ax.text(0.0, 9.15,
            "one of only two circuits that Strategy 5 alone fully solved",
            fontsize=8, color=MUTED, ha="left", va="top")

    attempt_block(ax, 7.10, "Attempt 1", A1, "execution_error", 14,
                  "name 'cirq' is not defined",
                  "circuit = cirq.Circuit()   # no import", ok=False)

    arrow(ax, (5.0, 6.10), (5.0, 4.85), color=CRIT, lw=1.4)
    ax.text(5.25, 5.48, "repair prompt injects  execution_error",
            fontsize=7.8, color=CRIT, ha="left", va="center")

    attempt_block(ax, 3.95, "Attempt 2", A2, "success", 100, "",
                  "+ import cirq", ok=True)

    ax.add_patch(FancyBboxPatch(
        (0.0, 1.45), 10.0, 1.15,
        boxstyle="round,pad=0,rounding_size=0.12",
        facecolor=BLUE_SOFT, edgecolor=BLUE, linewidth=1.1, zorder=2))
    ax.text(5.0, 2.02,
            "One added line moved the circuit from 1/7 to 7/7 checks.",
            fontsize=8.4, color=INK, ha="center", va="center", zorder=3)


def main(stem="fig4_pipeline"):
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11.0, 5.6),
                                   gridspec_kw={"width_ratios": [1, 1.15]})
    panel_a(ax1)
    panel_b(ax2)
    fig.subplots_adjust(left=0.02, right=0.98, top=0.97, bottom=0.03,
                        wspace=0.10)
    fig.savefig(f"{stem}.pdf")
    fig.savefig(f"{stem}.png", dpi=200)
    print(f"wrote {stem}.pdf and {stem}.png")

# --- render ---
main()
```
